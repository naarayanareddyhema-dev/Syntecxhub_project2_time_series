 Import Required Libraries
Pandas is used for handling data and Matplotlib is used for creating charts.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

STEP 2: Create Dataset

 Create Sales Dataset
This dataset contains Date, Sales values, and Category information

```python
data = {
    'Date': pd.date_range(start='2024-01-01', periods=6, freq='ME'),
    'Sales': [2000, 2500, 2200, 3000, 3200, 3500],
    'Category': ['Electronics', 'Clothing', 'Groceries',
                 'Electronics', 'Clothing', 'Groceries']
}

df = pd.DataFrame(data)
df
```

STEP 3: Time Series Chart (Line Chart)

Time Series Chart
A line chart is used to visualize sales trends over time.


```python
plt.figure()
plt.plot(df['Date'], df['Sales'])
plt.title('Sales Over Time')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.show()
```

Output:explaination
The line chart shows sales trends over time. Sales initially increase, experience a slight drop in the middle period, and then show a continuous upward trend. This indicates overall growth in sales with minor fluctuations.

STEP 4: Category Chart – Bar Chart

Category Comparison Using Bar Chart
A bar chart is used to compare total sales across categories.

```python
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure()
category_sales.plot(kind='bar')
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

 Output:explaination
The bar chart compares total sales across different product categories. Clothing and Groceries show higher sales compared to Electronics, indicating stronger performance in these categories.

STEP 5: Category Chart – Pie Chart

Category Share Using Pie Chart
A pie chart shows the percentage contribution of each category.

```python
plt.figure()
category_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title('Category Share')
plt.ylabel('')
plt.show()
```

Output:explaination
The pie chart shows the percentage contribution of each category to total sales. Clothing and Groceries contribute the highest share at 34.8% each, while Electronics contributes 30.5%. Minor variation in total percentage is due to rounding.

STEP 6: Save Output Charts

 Save Charts as Images
The charts are saved as PNG files for submission.


```python
# Save Time Series Chart
plt.figure()
plt.plot(df['Date'], df['Sales'])
plt.title('Sales Over Time')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.savefig('time_series_chart.png')
plt.close()

# Save Bar Chart
plt.figure()
category_sales.plot(kind='bar')
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.savefig('category_bar_chart.png')
plt.close()

#Save Pie Chart
plt.figure()
category_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title('Category Share')
plt.ylabel('')
plt.savefig('category_pie_chart.png')
plt.close()

print("All charts have been saved successfully.")
```

 Output Summary:
->A line chart showing sales trends over time.
->A bar chart comparing sales across categories.
->A pie chart showing category-wise sales contribution.
